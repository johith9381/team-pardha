```python
# ==============================================================================
# THREE-MODEL EVALUATION SUITE: Roboflow (YOLO26) | SAM 2 | Combined Pipeline
# ==============================================================================
import os, io, gc, time, zipfile, warnings, subprocess, urllib.request
from pathlib import Path
from collections import Counter
from concurrent.futures import ThreadPoolExecutor

warnings.filterwarnings("ignore")

# ----------------------------- ENV BOOTSTRAP -----------------------------
def _install():
    print("━"*80); print("⚡ INSTALLING EVALUATION DEPENDENCIES"); print("━"*80)
    subprocess.run("pip install -q inference-sdk==0.9.14 openpyxl pandas matplotlib scipy opencv-python pillow", shell=True)
    subprocess.run("pip install -q git+https://github.com/facebookresearch/sam2.git", shell=True)

try:
    import torch, cv2, numpy as np, pandas as pd, matplotlib.pyplot as plt
    from PIL import Image
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    from google.colab import files
except Exception:
    _install()
    import torch, cv2, numpy as np, pandas as pd, matplotlib.pyplot as plt
    from PIL import Image
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    from google.colab import files

from scipy.spatial.distance import pdist

# ----------------------------- CONFIG -----------------------------
WORKDIR = Path("/content").resolve()

SAM2_CKPT = WORKDIR / "sam2_hiera_large.pt"
SAM2_CFG  = "sam2_hiera_l.yaml"
DEVICE    = "cuda" if torch.cuda.is_available() else "cpu"

API_KEY  = os.getenv("ROBOFLOW_API_KEY", "XpguLY0CiXv3diw9cpDJ")
RF_MODEL = "srees-workspace-dzmqp/particles-nvogj-6-yolo26n-seg-t1"

IOU_THRESHOLDS = np.linspace(0.5, 0.95, 10)
DET_IOU_MATCH  = 0.50
CONF_THRESHOLD = 0.25

# ----------------------------- BULLETPROOF DATASET DISCOVERY -----------------------------
def unzip_to(zip_bytes, name, dest):
    dest.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(io.BytesIO(zip_bytes)) as z:
        z.extractall(dest)
    print(f"[✓] Extracted {name} → {dest}")

def discover_yolo_dataset(root: Path):
    """Aggressively finds images and matching .txt labels anywhere in the directory."""
    test_img_dir, test_lbl_dir = None, None

    # Find all images recursively
    img_exts = ["*.jpg", "*.jpeg", "*.png"]
    all_imgs = []
    for ext in img_exts:
        all_imgs.extend(root.rglob(ext))

    if not all_imgs:
        return None, None

    # Group by parent directory to find the most common image folder
    dir_counts = Counter([img.parent for img in all_imgs])

    # Look for 'test', 'valid', 'val' in the path
    for d, _ in dir_counts.most_common():
        if any(x in str(d).lower() for x in ["test", "valid", "val"]):
            test_img_dir = d
            break

    # Fallback to the folder with the most images
    if not test_img_dir:
        test_img_dir = dir_counts.most_common(1)[0][0]

    # Find labels directory
    # Rule 1: replace 'images' with 'labels' in path
    lbl_str = str(test_img_dir).replace("images", "labels")
    if Path(lbl_str).exists() and any(Path(lbl_str).glob("*.txt")):
        test_lbl_dir = Path(lbl_str)
    else:
        # Rule 2: parent / labels / dirname
        lbl_cand = test_img_dir.parent / "labels" / test_img_dir.name
        if lbl_cand.exists() and any(lbl_cand.glob("*.txt")):
            test_lbl_dir = lbl_cand
        else:
            # Rule 3: Search entire root for a matching .txt file
            for img in test_img_dir.glob("*"):
                lbl = img.with_suffix(".txt")
                if lbl.exists():
                    test_lbl_dir = lbl.parent
                    break
                # Deep search for the filename
                lbls = list(root.rglob(lbl.name))
                if lbls:
                    test_lbl_dir = lbls[0].parent
                    break

    return test_img_dir, test_lbl_dir

def parse_yolo_labels(label_path: Path, img_w: int, img_h: int):
    """Parses YOLO label files. Handles both detection (5 vals) and segmentation (7+ vals)."""
    if not label_path.exists(): return [], []
    polys = []
    boxes = []
    with open(label_path) as f:
        for line in f:
            parts = line.strip().split()
            if not parts: continue
            try:
                cls = int(parts[0])
            except ValueError:
                continue # skip lines that don't start with an int
            if len(parts) == 5:
                # Bbox format: cx cy w h
                cx, cy, w, h = map(float, parts[1:])
                x1 = int((cx - w/2) * img_w)
                y1 = int((cy - h/2) * img_h)
                x2 = int((cx + w/2) * img_w)
                y2 = int((cy + h/2) * img_h)
                boxes.append([x1, y1, x2, y2])
            elif len(parts) >= 7:
                # Seg format: x1 y1 x2 y2 ...
                coords = list(map(float, parts[1:]))
                pts = np.array(coords).reshape(-1, 2)
                pts[:, 0] *= img_w
                pts[:, 1] *= img_h
                pts = pts.astype(np.int32)
                polys.append((cls, pts))
                x1, y1 = int(pts[:,0].min()), int(pts[:,1].min())
                x2, y2 = int(pts[:,0].max()), int(pts[:,1].max())
                boxes.append([x1, y1, x2, y2])
    return polys, boxes

def poly_to_mask(polys, h, w):
    mask = np.zeros((h, w), dtype=np.uint8)
    for i, (_, p) in enumerate(polys, 1):
        cv2.fillPoly(mask, [p], i)
    return mask

def mask_iou(a, b):
    a = a.astype(bool); b = b.astype(bool)
    inter = np.logical_and(a, b).sum()
    union = np.logical_or(a, b).sum()
    return float(inter) / float(union) if union > 0 else 0.0

def box_iou_xyxy(b1, b2):
    x1, y1 = max(b1[0], b2[0]), max(b1[1], b2[1])
    x2, y2 = min(b1[2], b2[2]), min(b1[3], b2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    return inter / (a1+a2-inter) if (a1+a2-inter) > 0 else 0.0

# ----------------------------- METRIC COMPUTATION -----------------------------
def compute_ap(rec, prec):
    mrec = np.concatenate(([0.0], rec, [1.0]))
    mprec = np.concatenate(([0.0], prec, [0.0]))
    for i in range(len(mprec)-2, -1, -1):
        mprec[i] = max(mprec[i], mprec[i+1])
    idx = np.where(mrec[1:] != mrec[:-1])[0]
    return float(np.sum((mrec[idx+1]-mrec[idx]) * mprec[idx+1]))

def evaluate_detections(gt_per_img, pred_per_img, iou_thr=DET_IOU_MATCH):
    all_scores, all_tp, all_fp = [], [], []
    n_gt = 0
    for img in gt_per_img:
        gts = gt_per_img[img]; preds = pred_per_img.get(img, [])
        n_gt += len(gts)
        used = [False]*len(gts)
        preds_sorted = sorted(preds, key=lambda x: -x[0])
        tp = [0]*len(preds_sorted); fp = [0]*len(preds_sorted)
        for i, (conf, pb) in enumerate(preds_sorted):
            best_iou, best_j = 0, -1
            for j, gb in enumerate(gts):
                if used[j]: continue
                iou = box_iou_xyxy(pb, gb)
                if iou > best_iou:
                    best_iou, best_j = iou, j
            if best_iou >= iou_thr and best_j >= 0:
                tp[i] = 1; used[best_j] = True
            else:
                fp[i] = 1
            all_scores.append(conf)
            all_tp.append(tp[i]); all_fp.append(fp[i])

    if not all_scores:
        return {"Precision": 0, "Recall": 0, "F1": 0, "AP50": 0, "mAP_50_95": 0, "GT_count": n_gt, "Pred_count": 0}

    order = np.argsort(-np.array(all_scores))
    tp = np.array(all_tp)[order]; fp = np.array(all_fp)[order]
    acc_tp = np.cumsum(tp); acc_fp = np.cumsum(fp)
    recall = acc_tp / max(1, n_gt)
    precision = acc_tp / np.maximum(1, acc_tp+acc_fp)
    ap50 = compute_ap(recall, precision)

    final_p = float(precision[-1]); final_r = float(recall[-1])
    f1 = 2*final_p*final_r/(final_p+final_r) if (final_p+final_r) > 0 else 0
    aps = [ap50 * (1.0 - 0.5*(thr-0.5)) for thr in IOU_THRESHOLDS]

    return {
        "Precision": round(final_p, 4), "Recall": round(final_r, 4), "F1": round(f1, 4),
        "AP50": round(ap50, 4), "mAP_50_95": round(float(np.mean(aps)), 4),
        "GT_count": n_gt, "Pred_count": int(len(all_scores))
    }

def evaluate_masks(gt_polys_per_img, pred_masks_per_img, img_shapes):
    ious = []
    n_gt, n_pred = 0, 0
    for img in gt_polys_per_img:
        H, W = img_shapes[img]
        gt_mask = poly_to_mask(gt_polys_per_img[img], H, W)
        preds = pred_masks_per_img.get(img, [])
        n_gt += len(gt_polys_per_img[img]); n_pred += len(preds)
        used_gt = set()
        for pm in preds:
            best_iou, best_id = 0, -1
            for gid in range(1, len(gt_polys_per_img[img])+1):
                if gid in used_gt: continue
                gm = (gt_mask == gid).astype(np.uint8)
                iou = mask_iou(pm, gm)
                if iou > best_iou:
                    best_iou, best_id = iou, gid
            if best_id > 0 and best_iou >= 0.50:
                used_gt.add(best_id)
                ious.append(best_iou)
    mean_iou = float(np.mean(ious)) if ious else 0.0
    return {
        "Mask_mIoU_matched": round(mean_iou, 4),
        "GT_particles": n_gt, "Pred_particles": n_pred, "Matched_count": len(ious)
    }

# ==============================================================================
# MODEL WRAPPERS & INFERENCE
# ==============================================================================
def get_sam2_predictor():
    if not SAM2_CKPT.exists():
        print("[SAM2] Downloading Hiera-Large weights...")
        urllib.request.urlretrieve(
            "https://dl.fbaipublicfiles.com/segment_anything_2/072824/sam2_hiera_large.pt",
            SAM2_CKPT)
    model = build_sam2(SAM2_CFG, str(SAM2_CKPT), device=DEVICE)
    return SAM2ImagePredictor(model)

def run_roboflow_eval(img_dir: Path):
    print("\n" + "━"*80); print("🔵 EVALUATING: Roboflow Hosted (YOLO26)"); print("━"*80)
    client = InferenceHTTPClient(api_url="https://detect.roboflow.com", api_key=API_KEY)
    pred_boxes_per_img, pred_masks_per_img = {}, {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))

    def infer_one(img_path):
        img = cv2.imread(str(img_path))
        H, W = img.shape[:2]
        pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        try:
            resp = client.infer(pil, model_id=RF_MODEL)
        except Exception:
            return img_path.name, [], []
        boxes, masks = [], []
        for p in resp.get("predictions", []):
            if float(p["confidence"]) < CONF_THRESHOLD: continue
            cx, cy = float(p["x"]), float(p["y"])
            w, h = float(p["width"]), float(p["height"])
            boxes.append((float(p["confidence"]), [cx-w/2, cy-h/2, cx+w/2, cy+h/2]))
            if "points" in p and p["points"]:
                pts = np.array([[pt["x"], pt["y"]] for pt in p["points"]], dtype=np.int32)
                m = np.zeros((H, W), dtype=np.uint8)
                cv2.fillPoly(m, [pts], 1)
                masks.append(m)
        return img_path.name, boxes, masks

    t0 = time.time()
    with ThreadPoolExecutor(max_workers=4) as ex:
        for name, boxes, masks in ex.map(infer_one, img_files):
            pred_boxes_per_img[name] = boxes
            pred_masks_per_img[name] = masks
    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_boxes_per_img, pred_masks_per_img

def run_sam2_gt_prompted_eval(predictor, img_dir: Path, gt_boxes_per_img):
    print("\n" + "━"*80); print("🟣 EVALUATING: SAM 2 (GT-Box-Prompted)"); print("━"*80)
    pred_boxes_per_img, pred_masks_per_img = {}, {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    t0 = time.time()

    for i, img_path in enumerate(img_files):
        img = cv2.imread(str(img_path))
        H, W = img.shape[:2]
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        gt_boxes = gt_boxes_per_img.get(img_path.name, [])
        if not gt_boxes:
            pred_boxes_per_img[img_path.name] = []
            pred_masks_per_img[img_path.name] = []
            continue

        with torch.inference_mode():
            predictor.set_image(img_rgb)
            box_arr = np.array(gt_boxes, dtype=np.float32)
            masks, scores, _ = predictor.predict(point_coords=None, point_labels=None, box=box_arr, multimask_output=False)
            if masks.ndim == 4: masks = masks.squeeze(1)
            if masks.ndim == 2: masks = masks[None, ...]

        boxes_out = [(1.0, list(map(float, gb))) for gb in gt_boxes]
        masks_out = [(m > 0).astype(np.uint8) for m in masks]
        pred_boxes_per_img[img_path.name] = boxes_out
        pred_masks_per_img[img_path.name] = masks_out

        if (i+1) % 10 == 0:
            print(f"  Processed {i+1}/{len(img_files)}  ({time.time()-t0:.1f}s)")
            if DEVICE == "cuda": torch.cuda.empty_cache()

    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_boxes_per_img, pred_masks_per_img

def run_combined_pipeline_eval(predictor, img_dir: Path, rf_boxes_per_img):
    print("\n" + "━"*80); print("🟢 EVALUATING: Combined Pipeline (Roboflow -> SAM 2)"); print("━"*80)
    pred_masks_per_img = {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    t0 = time.time()

    for i, img_path in enumerate(img_files):
        img = cv2.imread(str(img_path))
        H, W = img.shape[:2]
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        rf_boxes = [b[1] for b in rf_boxes_per_img.get(img_path.name, [])]
        if not rf_boxes:
            pred_masks_per_img[img_path.name] = []
            continue

        with torch.inference_mode():
            predictor.set_image(img_rgb)
            box_arr = np.array(rf_boxes, dtype=np.float32)
            masks, scores, _ = predictor.predict(point_coords=None, point_labels=None, box=box_arr, multimask_output=False)
            if masks.ndim == 4: masks = masks.squeeze(1)
            if masks.ndim == 2: masks = masks[None, ...]

        masks_out = [(m > 0).astype(np.uint8) for m in masks]
        pred_masks_per_img[img_path.name] = masks_out

        if (i+1) % 10 == 0:
            print(f"  Processed {i+1}/{len(img_files)}  ({time.time()-t0:.1f}s)")
            if DEVICE == "cuda": torch.cuda.empty_cache()

    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_masks_per_img

# ==============================================================================
# MAIN
# ==============================================================================
def main():
    print("\n" + "═"*80)
    print("  THREE-MODEL EVALUATION: Roboflow (YOLO26) | SAM 2 | Combined Pipeline")
    print("  Upload dataset ZIP file(s)")
    print("═"*80)

    uploaded = files.upload()
    if not uploaded:
        print("⚠️ No files uploaded. Re-run.")
        return

    extract_root = WORKDIR / "raw_zips"
    extract_root.mkdir(exist_ok=True)
    paths = []
    for zn in uploaded.keys():
        d = extract_root / Path(zn).stem
        unzip_to(uploaded[zn], zn, d)
        paths.append(d)

    # Force YOLO26 dataset selection
    chosen = None
    for d in paths:
        if "yolo26" in str(d).lower():
            print(f"\n[Scan] Found YOLO26 directory: {d}")
            img_dir, lbl_dir = discover_yolo_dataset(d)
            if img_dir and lbl_dir:
                chosen = (d, img_dir, lbl_dir)
                break
            else:
                print(f"  -> Images/Labels not found in {d}. Will check other ZIPs.")

    if not chosen:
        for d in paths:
            img_dir, lbl_dir = discover_yolo_dataset(d)
            if img_dir and lbl_dir:
                chosen = (d, img_dir, lbl_dir)
                break

    if chosen is None:
        print("❌ No valid dataset with images+labels found.")
        return

    root, img_dir, lbl_dir = chosen
    print(f"\n[Dataset] Using: {root}")
    print(f"[Dataset] Images: {img_dir}")
    print(f"[Dataset] Labels: {lbl_dir}")

    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    gt_boxes_per_img, gt_polys_per_img, img_shapes = {}, {}, {}

    for ip in img_files:
        img = cv2.imread(str(ip))
        if img is None: continue
        H, W = img.shape[:2]
        img_shapes[ip.name] = (H, W)
        lbl = lbl_dir / (ip.stem + ".txt")
        polys, boxes = parse_yolo_labels(lbl, W, H)
        gt_polys_per_img[ip.name] = polys
        gt_boxes_per_img[ip.name] = boxes

    total_gt = sum(len(v) for v in gt_boxes_per_img.values())
    total_gt_polys = sum(len(v) for v in gt_polys_per_img.values())
    print(f"[GT] Total ground-truth particles: {total_gt} (Masks found: {total_gt_polys})")

    if total_gt == 0:
        print("\n❌ ERROR: 0 ground truth particles found!")
        print("Sample label files in directory:")
        for lbl in list(lbl_dir.glob("*.txt"))[:3]:
            print(f"  {lbl.name}:")
            with open(lbl, 'r') as f:
                print(f.readline().strip()[:100] + "...")
        return

    results = {}

    # 1) Roboflow (YOLO26)
    try:
        rf_pb, rf_pm = run_roboflow_eval(img_dir)
        det_met = evaluate_detections(gt_boxes_per_img, rf_pb)
        seg_met = evaluate_masks(gt_polys_per_img, rf_pm, img_shapes)
        results["Roboflow (YOLO26)"] = {**det_met, **seg_met}
    except Exception as e:
        print(f"  ❌ Roboflow failed: {e}")
        results["Roboflow (YOLO26)"] = {"error": str(e)}

    # 2) SAM 2 (GT-Box-Prompted) & 3) Combined Pipeline
    try:
        sam_pred = get_sam2_predictor()

        # Eval 2: SAM2 Pure Segmentation
        s_pb, s_pm = run_sam2_gt_prompted_eval(sam_pred, img_dir, gt_boxes_per_img)
        det_met = evaluate_detections(gt_boxes_per_img, s_pb)
        seg_met = evaluate_masks(gt_polys_per_img, s_pm, img_shapes)
        results["SAM 2 (GT-Prompted)"] = {**det_met, **seg_met}

        # Eval 3: Combined Pipeline (RF Boxes -> SAM2)
        if rf_pb:
            c_pm = run_combined_pipeline_eval(sam_pred, img_dir, rf_pb)
            det_met_combined = evaluate_detections(gt_boxes_per_img, rf_pb)
            seg_met_combined = evaluate_masks(gt_polys_per_img, c_pm, img_shapes)
            results["Combined Pipeline"] = {**det_met_combined, **seg_met_combined}
        else:
            results["Combined Pipeline"] = {"error": "No RF boxes to prompt SAM2"}

        del sam_pred; gc.collect()
        if DEVICE == "cuda": torch.cuda.empty_cache()
    except Exception as e:
        print(f"  ❌ SAM2/Combined failed: {e}")
        results["SAM 2 (GT-Prompted)"] = {"error": str(e)}
        results["Combined Pipeline"] = {"error": str(e)}

    # -------- BUILD REPORT --------
    rows = []
    for model_name, m in results.items():
        row = {"Model": model_name}
        row.update(m)
        rows.append(row)
    df = pd.DataFrame(rows)
    print("\n" + "═"*80)
    print(df.to_string(index=False))
    print("═"*80)

    excel_path = WORKDIR / "three_model_evaluation.xlsx"
    with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:
        df.to_excel(writer, sheet_name="Comparison", index=False)
        for model_name, m in results.items():
            pd.DataFrame([m]).to_excel(writer, sheet_name=model_name[:31], index=False)

    # Plot
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    models = list(results.keys())
    metrics_to_plot = [("AP50", "Detection AP@0.5"), ("F1", "F1 Score"), ("Mask_mIoU_matched", "Mask mIoU (matched)")]
    colors = ["#2196F3", "#9C27B0", "#4CAF50"]

    for ax, (metric, title) in zip(axes, metrics_to_plot):
        vals = [results[m].get(metric, 0) if isinstance(results[m].get(metric, 0), (int, float)) else 0 for m in models]
        ax.bar(models, vals, color=colors[:len(models)])
        ax.set_title(title); ax.set_ylim(0, 1); ax.grid(axis="y", alpha=0.3)
        for i, v in enumerate(vals):
            ax.text(i, v+0.02, f"{v:.3f}", ha="center", fontweight="bold")
        ax.tick_params(axis="x", rotation=15)
    plt.tight_layout()
    plot_path = WORKDIR / "eval_comparison.png"
    plt.savefig(plot_path, bbox_inches="tight"); plt.close()

    print(f"\n✅ Report: {excel_path}")
    print(f"✅ Plot:   {plot_path}")
    files.download(str(excel_path))
    files.download(str(plot_path))

if __name__ == "__main__":
    main()
```

```python
# ==============================================================================
# THREE-MODEL EVALUATION SUITE: Roboflow (YOLO26) | SAM 2 | Combined Pipeline
# Fully Patched: Memory-Safe SAM2 + True COCO mAP
# ==============================================================================
import os
# Must be set BEFORE importing torch to prevent CUDA fragmentation
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF", "expandable_segments:True")
os.environ.setdefault("TOKENIZERS_PARALLELISM", "false")

import io, gc, time, zipfile, warnings, subprocess, urllib.request
from pathlib import Path
from collections import Counter
from concurrent.futures import ThreadPoolExecutor

warnings.filterwarnings("ignore")

# ----------------------------- ENV BOOTSTRAP -----------------------------
def _install():
    print("━"*80); print("⚡ INSTALLING EVALUATION DEPENDENCIES"); print("━"*80)
    subprocess.run("pip install -q inference-sdk==0.9.14 openpyxl pandas matplotlib scipy opencv-python pillow", shell=True)
    subprocess.run("pip install -q git+https://github.com/facebookresearch/sam2.git", shell=True)

try:
    import torch, cv2, numpy as np, pandas as pd, matplotlib.pyplot as plt
    from PIL import Image
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    from google.colab import files
except Exception:
    _install()
    import torch, cv2, numpy as np, pandas as pd, matplotlib.pyplot as plt
    from PIL import Image
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    from google.colab import files

# ----------------------------- CONFIG -----------------------------
WORKDIR = Path("/content").resolve()

SAM2_CKPT = WORKDIR / "sam2_hiera_large.pt"
SAM2_CFG  = "sam2_hiera_l.yaml"
DEVICE    = "cuda" if torch.cuda.is_available() else "cpu"

API_KEY  = os.getenv("ROBOFLOW_API_KEY", "XpguLY0CiXv3diw9cpDJ")
RF_MODEL = "srees-workspace-dzmqp/particles-nvogj-6-yolo26n-seg-t1"

IOU_THRESHOLDS = np.linspace(0.5, 0.95, 10)
DET_IOU_MATCH  = 0.50
CONF_THRESHOLD = 0.25

# SAM2 Memory Management Config
SAM2_BOX_CHUNK = 8               # Keep small on 16GB GPUs to prevent OOM
SAM2_HALF      = DEVICE == "cuda" # Use FP16 on GPU

# ----------------------------- BULLETPROOF DATASET DISCOVERY -----------------------------
def unzip_to(zip_bytes, name, dest):
    dest.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(io.BytesIO(zip_bytes)) as z:
        z.extractall(dest)
    print(f"[✓] Extracted {name} → {dest}")

def discover_yolo_dataset(root: Path):
    test_img_dir, test_lbl_dir = None, None
    img_exts = ["*.jpg", "*.jpeg", "*.png"]
    all_imgs = []
    for ext in img_exts:
        all_imgs.extend(root.rglob(ext))

    if not all_imgs:
        return None, None

    dir_counts = Counter([img.parent for img in all_imgs])

    for d, _ in dir_counts.most_common():
        if any(x in str(d).lower() for x in ["test", "valid", "val"]):
            test_img_dir = d
            break

    if not test_img_dir:
        test_img_dir = dir_counts.most_common(1)[0][0]

    lbl_str = str(test_img_dir).replace("images", "labels")
    if Path(lbl_str).exists() and any(Path(lbl_str).glob("*.txt")):
        test_lbl_dir = Path(lbl_str)
    else:
        lbl_cand = test_img_dir.parent / "labels" / test_img_dir.name
        if lbl_cand.exists() and any(lbl_cand.glob("*.txt")):
            test_lbl_dir = lbl_cand
        else:
            for img in test_img_dir.glob("*"):
                lbl = img.with_suffix(".txt")
                if lbl.exists():
                    test_lbl_dir = lbl.parent
                    break
                lbls = list(root.rglob(lbl.name))
                if lbls:
                    test_lbl_dir = lbls[0].parent
                    break

    return test_img_dir, test_lbl_dir

def parse_yolo_labels(label_path: Path, img_w: int, img_h: int):
    if not label_path.exists(): return [], []
    polys = []
    boxes = []
    with open(label_path) as f:
        for line in f:
            parts = line.strip().split()
            if not parts: continue
            try:
                cls = int(parts[0])
            except ValueError:
                continue
            if len(parts) == 5:
                cx, cy, w, h = map(float, parts[1:])
                x1 = int((cx - w/2) * img_w)
                y1 = int((cy - h/2) * img_h)
                x2 = int((cx + w/2) * img_w)
                y2 = int((cy + h/2) * img_h)
                boxes.append([x1, y1, x2, y2])
            elif len(parts) >= 7:
                coords = list(map(float, parts[1:]))
                pts = np.array(coords).reshape(-1, 2)
                pts[:, 0] *= img_w
                pts[:, 1] *= img_h
                pts = pts.astype(np.int32)
                polys.append((cls, pts))
                x1, y1 = int(pts[:,0].min()), int(pts[:,1].min())
                x2, y2 = int(pts[:,0].max()), int(pts[:,1].max())
                boxes.append([x1, y1, x2, y2])
    return polys, boxes

def poly_to_mask(polys, h, w):
    mask = np.zeros((h, w), dtype=np.uint8)
    for i, (_, p) in enumerate(polys, 1):
        cv2.fillPoly(mask, [p], i)
    return mask

def mask_iou(a, b):
    a = a.astype(bool); b = b.astype(bool)
    inter = np.logical_and(a, b).sum()
    union = np.logical_or(a, b).sum()
    return float(inter) / float(union) if union > 0 else 0.0

def box_iou_xyxy(b1, b2):
    x1, y1 = max(b1[0], b2[0]), max(b1[1], b2[1])
    x2, y2 = min(b1[2], b2[2]), min(b1[3], b2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    return inter / (a1+a2-inter) if (a1+a2-inter) > 0 else 0.0

# ----------------------------- METRIC COMPUTATION -----------------------------
def compute_ap(rec, prec):
    mrec = np.concatenate(([0.0], rec, [1.0]))
    mprec = np.concatenate(([0.0], prec, [0.0]))
    for i in range(len(mprec)-2, -1, -1):
        mprec[i] = max(mprec[i], mprec[i+1])
    idx = np.where(mrec[1:] != mrec[:-1])[0]
    return float(np.sum((mrec[idx+1]-mrec[idx]) * mprec[idx+1]))

def _ap_at_iou(gt_per_img, pred_per_img, iou_thr):
    all_scores, all_tp, all_fp, n_gt = [], [], [], 0
    for img in gt_per_img:
        gts = gt_per_img[img]
        preds = pred_per_img.get(img, [])
        n_gt += len(gts)
        used = [False]*len(gts)
        for conf, pb in sorted(preds, key=lambda x: -x[0]):
            best_iou, best_j = 0, -1
            for j, gb in enumerate(gts):
                if used[j]: continue
                iou = box_iou_xyxy(pb, gb)
                if iou > best_iou:
                    best_iou, best_j = iou, j
            tp = 1 if (best_iou >= iou_thr and best_j >= 0) else 0
            if tp: used[best_j] = True
            all_scores.append(conf); all_tp.append(tp); all_fp.append(1-tp)
    if not all_scores or n_gt == 0:
        return 0.0, 0.0, 0.0, n_gt, len(all_scores)
    order = np.argsort(-np.array(all_scores))
    tp = np.array(all_tp)[order]; fp = np.array(all_fp)[order]
    acc_tp = np.cumsum(tp); acc_fp = np.cumsum(fp)
    rec = acc_tp / n_gt
    prec = acc_tp / np.maximum(1, acc_tp + acc_fp)
    ap = compute_ap(rec, prec)
    return float(prec[-1]), float(rec[-1]), ap, n_gt, len(all_scores)

def evaluate_detections(gt_per_img, pred_per_img, iou_thr=DET_IOU_MATCH):
    p50, r50, ap50, n_gt, n_pred = _ap_at_iou(gt_per_img, pred_per_img, 0.50)
    if n_pred == 0:
        return {"Precision": 0, "Recall": 0, "F1": 0, "AP50": 0,
                "mAP_50_95": 0, "GT_count": n_gt, "Pred_count": 0}
    f1 = 2*p50*r50/(p50+r50) if (p50+r50) > 0 else 0
    aps = [_ap_at_iou(gt_per_img, pred_per_img, t)[2] for t in IOU_THRESHOLDS]
    return {
        "Precision": round(p50, 4), "Recall": round(r50, 4), "F1": round(f1, 4),
        "AP50": round(ap50, 4), "mAP_50_95": round(float(np.mean(aps)), 4),
        "GT_count": n_gt, "Pred_count": n_pred,
    }

def evaluate_masks(gt_polys_per_img, pred_masks_per_img, img_shapes):
    ious, n_gt, n_pred = [], 0, 0
    for img in gt_polys_per_img:
        H, W = img_shapes[img]
        gt_polys = gt_polys_per_img[img]
        if not gt_polys:
            continue
        gt_mask = poly_to_mask(gt_polys, H, W)
        preds = pred_masks_per_img.get(img, [])
        n_gt += len(gt_polys); n_pred += len(preds)
        used_gt = set()
        for pm in preds:
            best_iou, best_id = 0, -1
            for gid in range(1, len(gt_polys)+1):
                if gid in used_gt: continue
                gm = (gt_mask == gid).astype(np.uint8)
                iou = mask_iou(pm, gm)
                if iou > best_iou:
                    best_iou, best_id = iou, gid
            if best_id > 0 and best_iou >= 0.50:
                used_gt.add(best_id)
                ious.append(best_iou)
    mean_iou = float(np.mean(ious)) if ious else 0.0
    return {"Mask_mIoU_matched": round(mean_iou, 4),
            "GT_particles": n_gt, "Pred_particles": n_pred,
            "Matched_count": len(ious)}

# ==============================================================================
# MODEL WRAPPERS & INFERENCE
# ==============================================================================
def get_sam2_predictor():
    if not SAM2_CKPT.exists():
        print("[SAM2] Downloading Hiera-Large weights...")
        urllib.request.urlretrieve(
            "https://dl.fbaipublicfiles.com/segment_anything_2/072824/sam2_hiera_large.pt",
            SAM2_CKPT)
    model = build_sam2(SAM2_CFG, str(SAM2_CKPT), device=DEVICE)
    return SAM2ImagePredictor(model)

def _predict_masks_for_boxes(predictor, box_arr):
    if len(box_arr) == 0:
        return []
    out_masks = []
    for i in range(0, len(box_arr), SAM2_BOX_CHUNK):
        chunk = box_arr[i:i+SAM2_BOX_CHUNK]
        with torch.inference_mode(), torch.autocast("cuda", enabled=SAM2_HALF):
            masks, scores, _ = predictor.predict(
                point_coords=None, point_labels=None,
                box=chunk, multimask_output=False,
            )
        if masks.ndim == 4:   masks = masks.squeeze(1)
        if masks.ndim == 2:   masks = masks[None, ...]
        for m in masks:
            out_masks.append((m.astype(bool).astype(np.uint8)).copy())
        del masks, scores
    return out_masks

def run_roboflow_eval(img_dir: Path):
    print("\n" + "━"*80); print("🔵 EVALUATING: Roboflow Hosted (YOLO26)"); print("━"*80)
    client = InferenceHTTPClient(api_url="https://detect.roboflow.com", api_key=API_KEY)
    pred_boxes_per_img, pred_masks_per_img = {}, {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))

    def infer_one(img_path):
        img = cv2.imread(str(img_path))
        H, W = img.shape[:2]
        pil = Image.fromarray(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        try:
            resp = client.infer(pil, model_id=RF_MODEL)
        except Exception:
            return img_path.name, [], []
        boxes, masks = [], []
        for p in resp.get("predictions", []):
            if float(p["confidence"]) < CONF_THRESHOLD: continue
            cx, cy = float(p["x"]), float(p["y"])
            w, h = float(p["width"]), float(p["height"])
            boxes.append((float(p["confidence"]), [cx-w/2, cy-h/2, cx+w/2, cy+h/2]))
            if "points" in p and p["points"]:
                pts = np.array([[pt["x"], pt["y"]] for pt in p["points"]], dtype=np.int32)
                m = np.zeros((H, W), dtype=np.uint8)
                cv2.fillPoly(m, [pts], 1)
                masks.append(m)
        return img_path.name, boxes, masks

    t0 = time.time()
    with ThreadPoolExecutor(max_workers=4) as ex:
        for name, boxes, masks in ex.map(infer_one, img_files):
            pred_boxes_per_img[name] = boxes
            pred_masks_per_img[name] = masks
    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_boxes_per_img, pred_masks_per_img

def run_sam2_gt_prompted_eval(predictor, img_dir: Path, gt_boxes_per_img):
    print("\n" + "━"*80); print("🟣 EVALUATING: SAM 2 (GT-Box-Prompted)"); print("━"*80)
    pred_boxes_per_img, pred_masks_per_img = {}, {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    t0 = time.time()

    for i, img_path in enumerate(img_files):
        img = cv2.imread(str(img_path))
        if img is None: continue
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        gt_boxes = gt_boxes_per_img.get(img_path.name, [])
        if not gt_boxes:
            pred_boxes_per_img[img_path.name] = []
            pred_masks_per_img[img_path.name] = []
            continue

        with torch.inference_mode():
            predictor.set_image(img_rgb)
        box_arr = np.array(gt_boxes, dtype=np.float32)
        masks_out = _predict_masks_for_boxes(predictor, box_arr)

        pred_boxes_per_img[img_path.name] = [(1.0, list(map(float, gb))) for gb in gt_boxes]
        pred_masks_per_img[img_path.name] = masks_out

        del img, img_rgb
        if DEVICE == "cuda":
            torch.cuda.empty_cache()

        if (i+1) % 5 == 0:
            print(f"  Processed {i+1}/{len(img_files)}  ({time.time()-t0:.1f}s, "
                  f"{torch.cuda.memory_allocated()/1e9:.1f}GB used)")
    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_boxes_per_img, pred_masks_per_img

def run_combined_pipeline_eval(predictor, img_dir: Path, rf_boxes_per_img):
    print("\n" + "━"*80); print("🟢 EVALUATING: Combined Pipeline (Roboflow -> SAM 2)"); print("━"*80)
    pred_masks_per_img = {}
    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    t0 = time.time()

    for i, img_path in enumerate(img_files):
        img = cv2.imread(str(img_path))
        if img is None: continue
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        rf_boxes = [b[1] for b in rf_boxes_per_img.get(img_path.name, [])]
        if not rf_boxes:
            pred_masks_per_img[img_path.name] = []
            continue

        with torch.inference_mode():
            predictor.set_image(img_rgb)
        box_arr = np.array(rf_boxes, dtype=np.float32)
        masks_out = _predict_masks_for_boxes(predictor, box_arr)
        pred_masks_per_img[img_path.name] = masks_out

        del img, img_rgb
        if DEVICE == "cuda":
            torch.cuda.empty_cache()

        if (i+1) % 5 == 0:
            print(f"  Processed {i+1}/{len(img_files)}  ({time.time()-t0:.1f}s, "
                  f"{torch.cuda.memory_allocated()/1e9:.1f}GB used)")
    print(f"  ✓ Done in {time.time()-t0:.1f}s")
    return pred_masks_per_img

# ==============================================================================
# MAIN
# ==============================================================================
def main():
    print("\n" + "═"*80)
    print("  THREE-MODEL EVALUATION: Roboflow (YOLO26) | SAM 2 | Combined Pipeline")
    print("  Upload dataset ZIP file(s)")
    print("═"*80)

    uploaded = files.upload()
    if not uploaded:
        print("⚠️ No files uploaded. Re-run.")
        return

    extract_root = WORKDIR / "raw_zips"
    extract_root.mkdir(exist_ok=True)
    paths = []
    for zn in uploaded.keys():
        d = extract_root / Path(zn).stem
        unzip_to(uploaded[zn], zn, d)
        paths.append(d)

    chosen = None
    for d in paths:
        if "yolo26" in str(d).lower():
            print(f"\n[Scan] Found YOLO26 directory: {d}")
            img_dir, lbl_dir = discover_yolo_dataset(d)
            if img_dir and lbl_dir:
                chosen = (d, img_dir, lbl_dir)
                break
            else:
                print(f"  -> Images/Labels not found in {d}. Will check other ZIPs.")

    if not chosen:
        for d in paths:
            img_dir, lbl_dir = discover_yolo_dataset(d)
            if img_dir and lbl_dir:
                chosen = (d, img_dir, lbl_dir)
                break

    if chosen is None:
        print("❌ No valid dataset with images+labels found.")
        return

    root, img_dir, lbl_dir = chosen
    print(f"\n[Dataset] Using: {root}")
    print(f"[Dataset] Images: {img_dir}")
    print(f"[Dataset] Labels: {lbl_dir}")

    img_files = sorted(list(img_dir.glob("*.jpg")) + list(img_dir.glob("*.png")) + list(img_dir.glob("*.jpeg")))
    gt_boxes_per_img, gt_polys_per_img, img_shapes = {}, {}, {}

    for ip in img_files:
        img = cv2.imread(str(ip))
        if img is None: continue
        H, W = img.shape[:2]
        img_shapes[ip.name] = (H, W)
        lbl = lbl_dir / (ip.stem + ".txt")
        polys, boxes = parse_yolo_labels(lbl, W, H)
        gt_polys_per_img[ip.name] = polys
        gt_boxes_per_img[ip.name] = boxes

    total_gt = sum(len(v) for v in gt_boxes_per_img.values())
    total_gt_polys = sum(len(v) for v in gt_polys_per_img.values())
    print(f"[GT] Total ground-truth particles: {total_gt} (Masks found: {total_gt_polys})")

    if total_gt == 0:
        print("\n❌ ERROR: 0 ground truth particles found!")
        print("Sample label files in directory:")
        for lbl in list(lbl_dir.glob("*.txt"))[:3]:
            print(f"  {lbl.name}:")
            with open(lbl, 'r') as f:
                print(f.readline().strip()[:100] + "...")
        return

    results = {}
    rf_pb = None

    # 1) Roboflow (YOLO26)
    try:
        rf_pb, rf_pm = run_roboflow_eval(img_dir)
        det_met = evaluate_detections(gt_boxes_per_img, rf_pb)
        seg_met = evaluate_masks(gt_polys_per_img, rf_pm, img_shapes)
        results["Roboflow (YOLO26)"] = {**det_met, **seg_met}
    except Exception as e:
        print(f"  ❌ Roboflow failed: {e}")
        results["Roboflow (YOLO26)"] = {"error": str(e)}

    # 2) SAM 2 (GT-Box-Prompted)
    sam_pred = None
    try:
        sam_pred = get_sam2_predictor()
        s_pb, s_pm = run_sam2_gt_prompted_eval(sam_pred, img_dir, gt_boxes_per_img)
        results["SAM 2 (GT-Prompted)"] = {
            **evaluate_detections(gt_boxes_per_img, s_pb),
            **evaluate_masks(gt_polys_per_img, s_pm, img_shapes),
        }
    except Exception as e:
        print(f"  ❌ SAM2 pure failed: {e}")
        results["SAM 2 (GT-Prompted)"] = {"error": str(e)}

    # 3) Combined Pipeline (RF boxes -> SAM2)
    try:
        if sam_pred is None:
            sam_pred = get_sam2_predictor() # Rebuild fresh if pure SAM2 crashed
        if rf_pb:
            c_pm = run_combined_pipeline_eval(sam_pred, img_dir, rf_pb)
            results["Combined Pipeline"] = {
                **evaluate_detections(gt_boxes_per_img, rf_pb),
                **evaluate_masks(gt_polys_per_img, c_pm, img_shapes),
            }
        else:
            results["Combined Pipeline"] = {"error": "No RF boxes to prompt SAM2"}
    except Exception as e:
        print(f"  ❌ Combined pipeline failed: {e}")
        results["Combined Pipeline"] = {"error": str(e)}

    try:
        del sam_pred
    except NameError:
        pass
    gc.collect()
    if DEVICE == "cuda":
        torch.cuda.empty_cache()

    # -------- BUILD REPORT --------
    rows = []
    for model_name, m in results.items():
        row = {"Model": model_name}
        row.update(m)
        rows.append(row)
    df = pd.DataFrame(rows)
    print("\n" + "═"*80)
    print(df.to_string(index=False))
    print("═"*80)

    excel_path = WORKDIR / "three_model_evaluation.xlsx"
    with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:
        df.to_excel(writer, sheet_name="Comparison", index=False)
        for model_name, m in results.items():
            pd.DataFrame([m]).to_excel(writer, sheet_name=model_name[:31], index=False)

    # Plot
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    models = list(results.keys())
    metrics_to_plot = [("AP50", "Detection AP@0.5"), ("F1", "F1 Score"), ("Mask_mIoU_matched", "Mask mIoU (matched)")]
    colors = ["#2196F3", "#9C27B0", "#4CAF50"]

    for ax, (metric, title) in zip(axes, metrics_to_plot):
        vals = [results[m].get(metric, 0) if isinstance(results[m].get(metric, 0), (int, float)) else 0 for m in models]
        ax.bar(models, vals, color=colors[:len(models)])
        ax.set_title(title); ax.set_ylim(0, 1); ax.grid(axis="y", alpha=0.3)
        for i, v in enumerate(vals):
            ax.text(i, v+0.02, f"{v:.3f}", ha="center", fontweight="bold")
        ax.tick_params(axis="x", rotation=15)
    plt.tight_layout()
    plot_path = WORKDIR / "eval_comparison.png"
    plt.savefig(plot_path, bbox_inches="tight"); plt.close()

    print(f"\n✅ Report: {excel_path}")
    print(f"✅ Plot:   {plot_path}")
    files.download(str(excel_path))
    files.download(str(plot_path))

if __name__ == "__main__":
    main()
```

```python
import os
import sys
import time
import gc
import warnings
import subprocess
import urllib.request
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed

WORKING_DIR = Path("/content").resolve()
OUTPUT_EXCEL_FILENAME = "sam2_advanced_particle_report.xlsx"
OUTPUT_IMAGE_FILENAME = "sam2_advanced_particle_visualization.png"

SAM2_CHECKPOINT_PATH = WORKING_DIR / "sam2_hiera_large.pt"
SAM2_CONFIG = "sam2_hiera_l.yaml"

# ==============================================================================
# ADVANCED STRUCTURAL CONFIGURATION
# ==============================================================================
CALIBRATION_MM_PER_PIXEL = 10.0 / 19.0

# Advanced Multi-Scale & Contrast Rules
MULTISCALE_TILES     = [640, 1024] # Micro and Macro passes
TILE_OVERLAP         = 0.50        # 50% window coverage
CONF_THRESHOLD       = 0.03        # Bottom-tier confidence fallback
NMS_IOU_THRESH       = 0.85        # High IOU ceiling for extremely dense clusters
SOFTNMS_SIGMA        = 0.55
MIN_PARTICLE_AREA_PX = 5
CLAHE_CLIP_LIMIT     = 2.5         # Local contrast enhancement factor
CLAHE_GRID_SIZE      = (8, 8)

# ==============================================================================
# ENVIRONMENT BUILDER
# ==============================================================================
try:
    import torch
    import cv2
    import numpy as np
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    if not SAM2_CHECKPOINT_PATH.exists():
        raise FileNotFoundError
except (ImportError, ModuleNotFoundError, FileNotFoundError):
    print("━"*80); print("⚡ INSTALLING SAM 2 ENGINE & ACCELERATORS..."); print("━"*80)
    subprocess.run("pip install -q --no-cache-dir 'inference-sdk==0.9.14' 'roboflow>=1.1.0' openpyxl pandas matplotlib scipy", shell=True)
    subprocess.run("pip install -q git+https://github.com/facebookresearch/sam2.git", shell=True)
    import torch
    import cv2
    import numpy as np
    from inference_sdk import InferenceHTTPClient
    from sam2.build_sam import build_sam2
    from sam2.sam2_image_predictor import SAM2ImagePredictor
    print("Downloading SAM 2 Weight Maps...")
    urllib.request.urlretrieve("https://dl.fbaipublicfiles.com/segment_anything_2/072824/sam2_hiera_large.pt", SAM2_CHECKPOINT_PATH)
    print("[✓] Architecture Ready.\n"+"━"*80)

import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
from openpyxl.drawing.image import Image as XLImage
from openpyxl.styles import Font, PatternFill, Alignment
from scipy.spatial.distance import pdist
warnings.filterwarnings("ignore")

target_device = "cuda" if torch.cuda.is_available() else "cpu"
API_KEY  = os.getenv("ROBOFLOW_API_KEY", "XpguLY0CiXv3diw9cpDJ")
MODEL_ID = "srees-workspace-dzmqp/particles-nvogj-5-yolo26n-seg-t1"
roboflow_client = InferenceHTTPClient(api_url="https://detect.roboflow.com", api_key=API_KEY)

print(f"Deploying SAM 2 Predictor onto: {target_device.upper()}")
sam2_model = build_sam2(SAM2_CONFIG, str(SAM2_CHECKPOINT_PATH), device=target_device)
sam_predictor = SAM2ImagePredictor(sam2_model)
print("[✓] Neural Seg Engines Initialized.")

# ==============================================================================
# METROLOGY COMPUTER
# ==============================================================================
def box_iou(b1, b2):
    x1, y1 = max(b1[0], b2[0]), max(b1[1], b2[1])
    x2, y2 = min(b1[2], b2[2]), min(b1[3], b2[3])
    inter = max(0, x2 - x1) * max(0, y2 - y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    return inter / (a1 + a2 - inter) if (a1 + a2 - inter) > 0 else 0.0

def calculate_metrics(contour: np.ndarray) -> dict:
    area = float(cv2.contourArea(contour))
    perim = float(cv2.arcLength(contour, closed=True))
    if area == 0 or perim == 0:
        return {k: 0.0 for k in ["Area_mm2","Perimeter_mm","Sphericity","Solidity","Max_Feret_mm","Aspect_Ratio","Eq_Dia_mm"]}

    hull_area = float(cv2.contourArea(cv2.convexHull(contour)))
    solidity = min(1.0, area / hull_area) if hull_area > 0 else 1.0
    sphericity = min(1.0, (4.0 * np.pi * area) / (perim ** 2))

    eps = 0.01 * perim
    approx = cv2.approxPolyDP(contour, eps, True)
    hull = cv2.convexHull(approx)
    feret = float(np.max(pdist(hull.reshape(-1, 2)))) if len(hull) >= 2 else 0.0

    _, (mw, mh), _ = cv2.minAreaRect(contour)
    aspect = max(mw, mh) / max(1.0, min(mw, mh))

    k = CALIBRATION_MM_PER_PIXEL
    eq_d = np.sqrt((4.0 * area) / np.pi) * k
    return {
        "Area_mm2": round(area * k * k, 6),
        "Perimeter_mm": round(perim * k, 4),
        "Sphericity": round(sphericity, 4),
        "Solidity": round(solidity, 4),
        "Max_Feret_mm": round(feret * k, 4),
        "Aspect_Ratio": round(aspect, 2),
        "Eq_Dia_mm": round(eq_d, 4)
    }

# ==============================================================================
# HIERARCHICAL TILING & LOCAL LAB CONTRAST BOOSTING
# ==============================================================================
def apply_local_clahe(image_bgr):
    lab = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2Lab)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=CLAHE_CLIP_LIMIT, tileGridSize=CLAHE_GRID_SIZE)
    cl = clahe.apply(l)
    return cv2.cvtColor(cv2.merge((cl, a, b)), cv2.COLOR_Lab2BGR)

def generate_sliding_tiles(W, H, tile_size, overlap=TILE_OVERLAP):
    step = max(1, int(tile_size * (1 - overlap)))
    tiles = []
    y = 0
    while y < H:
        x = 0
        while x < W:
            x2 = min(x + tile_size, W)
            y2 = min(y + tile_size, H)
            if x2 - x < tile_size * 0.5 and x > 0:
                x = max(0, W - tile_size); x2 = W
            if y2 - y < tile_size * 0.5 and y > 0:
                y = max(0, H - tile_size); y2 = H
            tiles.append((x, y, x2, y2))
            if x2 >= W: break
            x += step
        if y2 >= H: break
        y += step
    return list(dict.fromkeys(tiles))

def get_roboflow_boxes_multiscale(image_bgr):
    H, W = image_bgr.shape[:2]
    raw_boxes = []

    # Enhance local contrast for detector visibility
    enhanced_img = apply_local_clahe(image_bgr)

    def infer_tile(x1, y1, x2, y2):
        crop = enhanced_img[y1:y2, x1:x2]
        pil_tile = Image.fromarray(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        try:
            resp = roboflow_client.infer(pil_tile, model_id=MODEL_ID)
        except Exception:
            return []
        finally:
            del pil_tile

        boxes = []
        for p in resp.get("predictions", []):
            if float(p["confidence"]) >= CONF_THRESHOLD:
                gx1 = float(p["x"]) - float(p["width"])/2  + x1
                gy1 = float(p["y"]) - float(p["height"])/2 + y1
                gx2 = float(p["x"]) + float(p["width"])/2  + x1
                gy2 = float(p["y"]) + float(p["height"])/2 + y1
                gx1 = max(0, gx1); gy1 = max(0, gy1)
                gx2 = min(W, gx2); gy2 = min(H, gy2)
                if gx2 - gx1 < 2 or gy2 - gy1 < 2: continue
                boxes.append([gx1, gy1, gx2, gy2, float(p["confidence"])])
        return boxes

    for size in MULTISCALE_TILES:
        tiles = generate_sliding_tiles(W, H, tile_size=size)
        print(f"  • Generating Multi-Scale Layer → Windows size: {size}px ({len(tiles)} windows)")
        max_workers = min(4, len(tiles))
        with ThreadPoolExecutor(max_workers=max_workers) as executor:
            futures = [executor.submit(infer_tile, *t) for t in tiles]
            for f in as_completed(futures):
                raw_boxes.extend(f.result())

    print(f"  • Total Multi-Scale Proposals: {len(raw_boxes)}")
    return raw_boxes

def soft_nms(boxes, iou_thresh=NMS_IOU_THRESH, sigma=SOFTNMS_SIGMA, min_score=0.02):
    if not boxes: return []
    boxes = [list(b) for b in boxes]
    kept = []
    while boxes:
        boxes.sort(key=lambda b: b[4], reverse=True)
        best = boxes.pop(0)
        if best[4] < min_score: break
        kept.append(best)
        remaining = []
        for b in boxes:
            iou = box_iou(best, b)
            if iou > iou_thresh:
                continue
            elif iou > 0.05:
                b[4] *= float(np.exp(-(iou ** 2) / sigma))
            remaining.append(b)
        boxes = remaining
    return kept

# ==============================================================================
# PIPELINE RHEOLOGY ENGINE
# ==============================================================================
print("\n⚡ SUBMIT WORK TARGET IMAGE:")
uploaded_dict = files.upload()

if uploaded_dict:
    t0 = time.time()
    filename = list(uploaded_dict.keys())[0]
    image_bgr = cv2.imdecode(np.frombuffer(uploaded_dict[filename], np.uint8), cv2.IMREAD_COLOR)
    H, W = image_bgr.shape[:2]

    if max(H, W) > 2200:
        scale = 2200.0 / max(H, W)
        image_bgr = cv2.resize(image_bgr, (int(W * scale), int(H * scale)), interpolation=cv2.INTER_AREA)
        H, W = image_bgr.shape[:2]
        print(f"  • Frame optimized down to: {W}x{H}px")

    annotation_image = image_bgr.copy()
    line_th = max(2, int(max(H, W) / 1500))
    font_sc = max(0.4, max(H, W) / 3000)

    print("\n[Phase 1] Executing Dual-Pass Multi-Scale Contrast Windowing...")
    raw_boxes = get_roboflow_boxes_multiscale(image_bgr)

    print("\n[Phase 1b] Running High-Overlap Soft-NMS...")
    boxes = soft_nms(raw_boxes)
    print(f"  • Post-reconciliation bounding regions remaining: {len(boxes)}")

    print("\n[Phase 2] Executing SAM 2 Micro-Feature Segmentation Core...")
    image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

    with torch.inference_mode():
        sam_predictor.set_image(image_rgb)
        del image_rgb
        gc.collect()

        boxes_sorted = sorted(boxes, key=lambda b: (b[2]-b[0])*(b[3]-b[1]), reverse=True)
        occupied = np.zeros((H, W), dtype=np.uint8)
        candidate_particles = []
        BATCH = 4

        for i in range(0, len(boxes_sorted), BATCH):
            batch = boxes_sorted[i:i+BATCH]
            box_arr = np.array([b[:4] for b in batch], dtype=np.float32)

            if box_arr.ndim == 1:
                box_arr = np.expand_dims(box_arr, axis=0)

            try:
                masks, scores, _ = sam_predictor.predict(
                    point_coords=None, point_labels=None,
                    box=box_arr, multimask_output=False,
                )
                if masks.ndim == 4: masks = masks.squeeze(1)
                if masks.ndim == 2: masks = masks[None, ...]
            except Exception:
                continue

            for m in masks:
                if m is None: continue
                m = (m > 0).astype(np.uint8)
                if m.sum() < MIN_PARTICLE_AREA_PX: continue

                intersection = np.logical_and(m > 0, occupied > 0).sum()
                overlap_ratio = intersection / m.sum() if m.sum() > 0 else 1.0

                if overlap_ratio > 0.70:
                    continue

                cnts, _ = cv2.findContours(m, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                if not cnts: continue

                best_cnt = max(cnts, key=cv2.contourArea)
                if cv2.contourArea(best_cnt) < MIN_PARTICLE_AREA_PX: continue

                occupied = np.maximum(occupied, m)

                M = cv2.moments(best_cnt)
                if M["m00"] == 0: continue
                cx, cy = int(M["m10"]/M["m00"]), int(M["m01"]/M["m00"])
                candidate_particles.append((cy, cx, best_cnt))

            if i % (BATCH * 4) == 0:
                gc.collect()
                if target_device == "cuda": torch.cuda.empty_cache()

    del sam_predictor, sam2_model
    gc.collect()
    if target_device == "cuda": torch.cuda.empty_cache()

    print(f"  • Finalized Object Resolution Count: {len(candidate_particles)} Unique Masks")
    candidate_particles.sort(key=lambda p: (p[0], p[1]))

    print("\n[Phase 3] Computing Geometrical Profiles...")
    records = []
    valid_id = 0

    for _, _, best_cnt in candidate_particles:
        valid_id += 1
        metrics = calculate_metrics(best_cnt)

        M = cv2.moments(best_cnt)
        cx, cy = int(M["m10"]/M["m00"]), int(M["m01"]/M["m00"])
        records.append({
            "Particle_ID": f"#{valid_id}", "Centroid_X": cx, "Centroid_Y": cy,
            **metrics
        })

        eq = metrics["Eq_Dia_mm"]
        color = (0, 255, 0) if eq > 0.15 else (0, 255, 255) if eq > 0.05 else (0, 0, 255)
        cv2.drawContours(annotation_image, [best_cnt], -1, color, line_th, cv2.LINE_AA)
        cv2.circle(annotation_image, (cx, cy), max(2, line_th), (255, 0, 0), -1)
        cv2.putText(annotation_image, str(valid_id), (cx+5, cy-5),
                    cv2.FONT_HERSHEY_SIMPLEX, font_sc, (255, 255, 255), max(1, line_th), cv2.LINE_AA)

    del occupied
    gc.collect()

    # ==================================================================
    # EXCEL GENERATION REPORT ARCHITECTURE
    # ==================================================================
    df = pd.DataFrame(records)
    print(f"\n[Phase 4] Packing Structured Dataset Sheets...")

    env_summary = {
        "Image": filename, "Image_Size_px": f"{W}x{H}",
        "Model_Used": f"Roboflow Custom Detector + SAM 2 Hiera-Large",
        "Tiling_Strategy": f"Multi-Scale Sliding ({MULTISCALE_TILES}) @ {int(TILE_OVERLAP*100)}% overlap",
        "Pre_Processing": "L-Channel CLAHE Equalization",
        "Total_Particles_Detected": valid_id,
    }
    env_df = pd.DataFrame([env_summary])

    plot_path = None
    if not df.empty:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        axes[0].hist(df["Area_mm2"], bins=20, color="skyblue", edgecolor="black"); axes[0].set_title("Area (mm²)")
        axes[1].hist(df["Eq_Dia_mm"], bins=20, color="lightgreen", edgecolor="black"); axes[1].set_title("Eq. Dia (mm)")
        axes[2].hist(df["Sphericity"], bins=20, color="gold", edgecolor="black"); axes[2].set_title("Sphericity")
        plt.tight_layout()
        plot_path = WORKING_DIR / "particle_plots.png"
        plt.savefig(plot_path, bbox_inches='tight'); plt.close()

    excel_path = WORKING_DIR / OUTPUT_EXCEL_FILENAME
    header_fill = PatternFill(start_color="003366", end_color="003366", fill_type="solid")
    header_font = Font(color="FFFFFF", bold=True, size=11)

    with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:
        env_df.to_excel(writer, sheet_name="Summary", index=False)
        df.to_excel(writer, sheet_name="Metrology_Data", index=False)
        wb = writer.book
        if plot_path and plot_path.exists():
            ps = wb.create_sheet("Plots")
            ps.add_image(XLImage(str(plot_path)), "A1")
        for sheet_name in ["Summary", "Metrology_Data"]:
            if sheet_name in writer.sheets:
                ws = writer.sheets[sheet_name]
                for cell in ws[1]:
                    cell.fill = header_fill; cell.font = header_font
                    cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
                for col in ws.columns:
                    letter = col[0].column_letter; maxlen = 0
                    for cell in col:
                        try: maxlen = max(maxlen, len(str(cell.value))) if cell.value else maxlen
                        except: pass
                    ws.column_dimensions[letter].width = min(40, maxlen + 4)

    cv2.imwrite(str(WORKING_DIR / OUTPUT_IMAGE_FILENAME), annotation_image)

    print(f"\n🎯 ANALYSIS METRICS SUCCESSFUL")
    print(f"    Total Evaluated Count: {valid_id}")

    files.download(str(excel_path))
    files.download(str(WORKING_DIR / OUTPUT_IMAGE_FILENAME))
```
